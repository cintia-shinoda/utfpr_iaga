## Questão 1

Faça um programa para ler um valor monetário (R$) de um produto e calcule um desconto fixo de 10%. Calcule também o valor final do produto. Apresente os valores como apresentado na coluna “Resultados”.

<img src="./images/atividade-avaliativa/q1.png" width = "75%" align =" center" alt="Questão 1">

```python
# Solução do autor:

vlr = float(input())

desc = vlr * (10 / 100)
vlrFinal = vlr - desc

print(f"Valor do Produto  = R$ {vlr:.2f}")
print(f"Valor do Desconto = R$ {desc:.2f}")
print(f"Valor Final       = R$ {vlrFinal:.2f}")
```

## Questão 2

Faça um programa que leia um valor inteiro do teclado (variável “item”) e a seguir implemente a rotina de busca sequencial: do primeiro elemento da Lista até encontrar; ou até o final da Lista e não encontrar. Mostre as posições juntas dos respectivos elementos da Lista “a” e, finalizando, com o resultado da busca: encontrou na posição tal; ou não encontrou “item” na Lista “a”. Observe o resultado final na coluna “Resultados”.

<img src="./images/atividade-avaliativa/q2.png" width = "75%" align =" center" alt="Questão 2">

```python
a = [10, 2, 7, 8, 5, 3, 22, 17, 18]

item = int(input())

# Mostra as posições juntas dos respectivos elementos no formato exato {0:10, 1:2, ...}
posicoes_elementos = ", ".join(f"{i}:{a[i]}" for i in range(len(a)))
print("{" + posicoes_elementos + "}")
print()  # Linha em branco conforme a coluna "Resultados"

# Rotina de busca sequencial: do primeiro elemento até encontrar ou até o final da lista
i = 0
encontrou = False

while i < len(a) and not encontrou:
    if a[i] == item:
        encontrou = True
    else:
        i += 1

# Resultado final da busca
if encontrou:
    print(f"Item: {item}, foi encontrado na posicao {i}.")
else:
    print(f'Item: {item}, "nao" foi encontrado.')
```

```python
# Solução do autor:

a = [10, 2, 7, 8, 5, 3, 22, 17, 18]

item = int(input())

achou = False

n = len(a)
print("{", end="")
for i in range(n):
    print(f"{i}:{a[i]}", end="")
    if (i != (n-1)):
        print(", ", end="")
print("}")

for i in range(n):
    if (a[i] == item):
        achou = True
        break
    
print()

if (achou):
    print(f"Item: {item}, foi encontrado na posicao {i}.")
else:
    print(f"Item: {item}, \"nao\" foi encontrado.")
```

## Questão 3

Faça um programa para ler um valor monetário (R$) de um produto e calcule um desconto de 10\% se o valor do produto for inferior a R$ 5.000,00; o desconto a ser calculado, em caso contrário, deverá ser de 15%. Calcule, também, o valor final do produto e apresente os valores como mostrado na coluna 'Resultados'.


 <img src="./images/atividade-avaliativa/q3.png" width="80%" align="center" />

```python
valor_produto = float(input())

if valor_produto < 5000:
    valor_desconto = valor_produto * 0.10
    porcentagem = 10
else:
    valor_desconto = valor_produto * 0.15
    porcentagem = 15

print(f"Valor do Produto        = R$ {valor_produto:.2f}")
print(f"Valor do Desconto ({porcentagem}%) = R$ {valor_desconto:.2f}")
print(f"Valor Final             = R$ {valor_produto - valor_desconto:.2f}")
```

```python
# Solução do autor:

vlr = float(input())

print(f"Valor do Produto        = R$ {vlr:.2f}")

if (vlr >= 5000.00):
    desc = vlr * (15 / 100)
    print(f"Valor do Desconto (15%) = R$ {desc:.2f}")
else:
    desc = vlr * (10 / 100)
    print(f"Valor do Desconto (10%) = R$ {desc:.2f}")
    
vlrFinal = vlr - desc
print(f"Valor Final             = R$ {vlrFinal:.2f}")
```

## Questão 4

Faça um programa para ler um número inteiro que representa o tamanho de uma matriz quadrada. Apresente esta matriz como mostrado na coluna “Resultados”. Observe que as bordas da matriz são feitas por asteriscos (*) e a parte central por símbolos de arroba (@).

<img src="./images/atividade-avaliativa/q4.png" width="40%" align="center" />

```python
n = int(input())

for i in range(n):
    if i == 0 or i == n - 1:
        print(" ".join(["*"] * n))
    else:
        print(" ".join(["*"] + ["@"] * (n - 2) + ["*"]))
```

```python
# Solução do autor:

n = int(input())

for i in range(n):
    for j in range(n):
        if (i == 0) or (i == (n-1)):
            print("*", end=" ")
        elif (j == 0) or (j == (n-1)):
            print("*", end=" ")
        else:
            print("@", end=" ")
    print()
```

## Questão 6

Faça um programa para ler uma String e apresente todos as palavras desta String. Leve em consideração que entre as palavras existem um, e somente um, caractere espaço. Veja, também, que o ponto final não faz parte da última palavra. Observe o resultado final na coluna “Resultados”.

<img src="./images/atividade-avaliativa/q6.png" width="80%" align="center" />

```python
texto = input()

# Se o último caractere for ponto final, remove-o via slicing
if len(texto) > 0 and texto[-1] == ".":
    texto = texto[:-1]

palavras = texto.split(" ")

print(f"Existem {len(palavras)} palavras, são elas:\n")

contador = 1
for palavra in palavras:
    print(f"{contador}a. palavra = {palavra}")
    contador += 1
```

```python
# Solução do autor:

s = input()

palavras = s.split(" ")
ultimaPalavra = len(palavras)-1
ultimoCaractere = len(palavras[ultimaPalavra])-1
if (palavras[ultimaPalavra][ultimoCaractere] == '.'):
    palavras[ultimaPalavra] = palavras[ultimaPalavra][:ultimoCaractere]

n = len(palavras)
print(f"Existem {n} palavras, são elas:\n")
for i in range(n):
    print(f"{i+1}a. palavra = {palavras[i]}")
```

## Questão 7

Faça um programa para ler um valor inteiro que representa o tamanho de uma Lista e a seguir atribua para as posiões pares desta lista o valor 0 (zero) e para as posições ímpares o valor 1 (um). Finalizando, mostre os valores da Lista como mostrado na coluna 'Resultados'.

<img src="./images/atividade-avaliativa/q7.png" width="80%" align="center" />

```python
n = int(input())

lista = []
for i in range(n):
    if i % 2 == 0:
        lista.append(0)
    else:
        lista.append(1)

# Formata a lista com chaves {0, 1, 0, ...} conforme a coluna "Resultados"
elementos = ", ".join(str(x) for x in lista)
print("{" + elementos + "}")
```

```python
# Solução do autor:

n = int(input())

x = [0] * n

for i in range(n):
    if ((i % 2) == 1):
        x[i] = 1

print("{", end="")
for i in range(n):
    print(x[i], end="")
    if (i != (n-1)):
        print(", ", end="")
print("}")
```

## Questão 8

Faça um programa para ler do teclado 7 (sete) valores inteiros e os armazene em uma Lista. Em seguida, deverão ser mostrados na tela do computador: todos os elementos da Lista e os menor e maior elementos desta Lista com as suas respectivas posições. Observe o resultado final na coluna “Resultados”.

<img src="./images/atividade-avaliativa/q8.png" width="50%" align="center" />

```python
x = [int(input()) for _ in range(7)]

for i in range(len(x)):
    print(f"x[{i}] = {x[i]}")

menor = min(x)
pos_menor = x.index(menor)

maior = max(x)
pos_maior = x.index(maior)

print()
print(f"Menor elemento, x[{pos_menor}] = {menor}")
print(f"Maior elemento, x[{pos_maior}] = {maior}")
```

```python
# Solução do autor:

n = 7
x = []
for i in range(n):
    x.append(int(input()))

posMenor = posMaior = 0
menor = x[0]
maior = x[0]
for i in range(1, n):
    if (x[i] < x[posMenor]):
        posMenor = i
        menor = x[i]
    elif (x[i] > x[posMaior]):
        posMaior = i
        maior = x[i]

for i in range(n):
    print(f"x[{i}] = {x[i]}")

print()
print(f"Menor elemento, x[{posMenor}] = {menor}")
print(f"Maior elemento, x[{posMaior}] = {maior}")
```

## Questão 9

Faça um programa para ler um valor inteiro (variável “a”) e a seguir construa uma Lista com 10 (dez) posições (variável “x”) com o resultado da seguinte expressão:
x[i] = a + i; para todo “i” variando de 0 até 9.

Observe o resultado final na coluna “Resultados”.

<img src="./images/atividade-avaliativa/q9.png" width="40%" align="center" />

```python
a = int(input())

x = [0] * 10

for i in range(10):
    x[i] = a + i
    print(f"x[{i}] = {a} + {i} = {x[i]}")
```

```python
# Solução do autor:

a = int(input())

n = 10
x = []
for i in range(n):
    x.append(a + i)

for i in range(n):
    print(f"x[{i}] = {a} + {i} = {x[i]}")
```

## Questão 10

Faça um programa para ler um valor inteiro que representa o tamanho (variável “n”) de uma primeira Lista e a seguir leia todos os “n” elementos desta Lista. E, também, construa uma segunda lista onde seus elementos são assim definidos: a) se o elemento da primeira lista for um valor negativo, armazenar na segunda lista o valor -1 (menos um); b) se o elemento da primeira lista for o valor neutro (zero), armazenar na segunda lista o valor 0 (zero); e, c) se o elemento da primeira lista for um valor positivo, armazenar na segunda lista o valor 1 (um). Finalizando, mostre os valores das Listas como mostrado na coluna 'Resultados'.

<img src="./images/atividade-avaliativa/q10.png" width="60%" align="center" />

```python
n = int(input())

lista1 = [int(input()) for _ in range(n)]
lista2 = [-1 if x < 0 else (0 if x == 0 else 1) for x in lista1]

print(lista1)
print(lista2)
```

```python
# Solução do autor:

n = int(input())

a = []
for i in range(n):
    a.append(int(input()))

b = []
for i in range(n):
    if (a[i] < 0):
        b.append(-1)
    elif (a[i] == 0):
        b.append(0)
    else:
        b.append(1)

print(a)
print(b)
```
