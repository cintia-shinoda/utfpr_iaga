## Questão 1

Fazer um programa que leia dois números inteiros e calcule a média aritmética simples destes números imprimindo o resultado com uma casa decimal.

```python
n1 = int(input())
n2 = int(input())

media = (n1 + n2) / 2
print(f"Media = {media:.1f}")
```

## Questão 2

Escreva um programa que leia o lado de quadrado e calcule os valores de:

a) área = $lado^{2}$

b) perímetro = $4 \times lado$

c) comprimento da diagonal = $lado \sqrt{2}$

```python
import math

lado = float(input())

area = lado ** 2
perimetro = 4 * lado
diagonal = lado * math.sqrt(2)

print(f"Area.....: {area:.2f}")
print(f"Perimetro: {perimetro:.2f}")
print(f"Diagonal.: {diagonal:.2f}")
```

## Questão 3

Ler um número inteiro e imprimir seu sucessor e seu antecessor.

```python
numero = int(input())

antecessor = numero - 1
sucessor = numero + 1

print(f"antecessor: {antecessor}")
print(f"numero....: {numero}")
print(f"sucessor..: {sucessor}")
```

## Questão 4

Implementar um programa que obtenha o valor do salário atual de um funcionário e um percentual de reajuste. A seguir, calcule e escreva o valor atual do salário, o valor do reajuste e o novo valor do salário acrescido do valor do reajuste. Valores com duas casas decimais.

```python
salario_atual = float(input())
percentual_reajuste = float(input())

valor_reajuste = salario_atual * (percentual_reajuste / 100)
salario_novo = salario_atual + valor_reajuste

print(f"Salario atual....: R$ {salario_atual:.2f}")
print(f"Valor do reajuste: R$ {valor_reajuste:.2f} ({percentual_reajuste:.2f}%)")
print(f"Salario novo.....: R$ {salario_novo:.2f}")
```

## Questão 5

Fazer um programa que leia dois números inteiros e a seguir realize os seguintes cálculos com esses valores: soma, subtração, multiplicação e divisão (inteira e real). Mostre no terminal de vídeo os valores calculados. A divisão real deverá ser apresentada com duas casas decimais.

```python
n1 = int(input())
n2 = int(input())

soma = n1 + n2
subtracao = n1 - n2
multiplicacao = n1 * n2
divisao_inteira = n1 // n2
divisao_real = n1 / n2

print(f"{n1} + {n2} = {soma}")
print(f"{n1} - {n2} = {subtracao}")
print(f"{n1} X {n2} = {multiplicacao}")
print(f"{n1} / {n2} = {divisao_inteira}")
print(f"{n1} / {n2} = {divisao_real:.2f}")
```
